## Problem 5

```python
class Every():
  def one(self): return type(self).__name__ + self.two();
  def two(self): return "thing"

class Some(Every):
  def two(self): return "where"

class Any(Some):
  pass

a = Some()
print(a.one()) # 1. Somewhere

b = Every()
print(b.one()) # 2. Everything

c = Any()
print(c.one()) # 3. Anywhere
```

## Problem 7

```python
import numpy as np

a = np.array([4, 2, 6, 5, 3, 8, 1, 7])
print(a[-2:3:-1]) # [1] prints [1 8 3]

b = a.reshape(2, 4)
print(b[0:2,1:3]) # [2] prints [[2 6] [8 1]]

c = a.reshape(2, 2, 2)
print(c[0,:,:])   # [3] prints [[4 2] [6 5]]

print(c[:,1,:].T) # [4] prints [[6 1] [5 7]]

print(c[:,:,1])   # [5] prints [[2 5] [8 7]]
```

## Problem 8

```python
class LinkedList:
  def __init__(self, data, next=None):
    self.data = data
    self.next = next

  def show(self):
    result = ""
    temp = self
    while temp:
      result += " " + str(temp.data) + " "
      temp = temp.next
    return result

  def insert(self, value):
    return LinkedList(value, self)

a = LinkedList(5, None)
print(a.show()) # 5

a = a.insert(4)
print(a.show()) # 4 5

a = a.insert(3)
print(a.show()) # 3 4 5

a = a.insert(2)
print(a.show()) # 2 3 4 5

a = a.insert(1)
print(a.show()) # 1 2 3 4 5
```
